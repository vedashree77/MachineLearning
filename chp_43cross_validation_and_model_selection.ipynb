```python
from sklearn.model_selection import cross_val_score, KFold, StratifiedKFold
from sklearn.linear_model import LogisticRegression 
from sklearn.datasets import load_iris 
from sklearn.preprocessing import StandardScaler
import pandas as pd
```

```python
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler

iris = load_iris()
x=iris.data
y=iris.target

scaler =StandardScaler()
x_scaled=scaler.fit_transform(x)
```

```python
x_scaled_df=pd.DataFrame(
    x_scaled,
    columns=iris.feature_names
)
x_scaled_df["target"]=y
x_scaled_df.head()
```

```python
model = LogisticRegression(max_iter = 1000)
```

```python
kf = KFold(n_splits = 5, shuffle = True, random_state = 42)
scores = cross_val_score(model, x_scaled, y, cv = kf)
print("Accuracy for each fold: ", scores)
print("Average Accuracy:", scores.mean())
```

```python
skf=StratifiedKFold(n_splits=5,shuffle=True,random_state=1)
scores=cross_val_score(model,x_scaled,y,cv=skf)
print("Stratified Accuracy Scores:",scores)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x_scaled,y,test_size=0.3,random_state=1)
model.fit(x_train,y_train)
print("Single split accuracy:",model.score(x_test,y_test))
```
